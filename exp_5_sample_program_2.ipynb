```python
#Taking input for the first set
set1 = set(map(int,input("Enter elements of first set(space-separated):").split()))

#Taking input for the second set
set2 = set(map(int,input("Enter elements of second set(space-seperated):").split()))

#Union of two sets
union_set = set1|set2 # or set 1.union(set2)

#Intersection of two sets
intersection_set = set1 & set2 # or set1.intersection

#Display the results
print("Union of the sets:",union_set)
print("Intersection of the sets:",intersection_set)
```
