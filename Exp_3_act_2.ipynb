```python
def emi(P,R,Y):
  r = R / (12 * 100)
  n = Y * 12

  if r == 0:
    return P / n
  e = (P * r * (1 + r) ** n) / ((1 + r) ** n - 1)
  return e
P = float(input("Enter loan ammount: "))
R = float(input("Enter annual interest rate: "))
Y = float(input("Enter years: "))
print("EMI =", emi(P,R,Y))
```
